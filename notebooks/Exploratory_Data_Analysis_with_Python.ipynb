# Analysis of Variance
## Drive Wheels

```python
grouped_wheels = df.groupby('drive-wheels')
```

```python
f_val, p_val = stats.f_oneway(
    grouped_wheels.get_group('fwd')['price'],
    grouped_wheels.get_group('rwd')['price'],
    grouped_wheels.get_group('4wd')['price']
)

print("ANOVA results: F =", f_val, ", P =", p_val)
```

## fwd and rwd

```python
f_val, p_val = stats.f_oneway(
    grouped_wheels.get_group('fwd')['price'], 
    grouped_wheels.get_group('rwd')['price']
)  
 
print( "ANOVA results: F=", f_val, ", P =", p_val )
```

## 4wd and rwd

```python
f_val, p_val = stats.f_oneway(
    grouped_wheels.get_group('4wd')['price'], 
    grouped_wheels.get_group('rwd')['price']
)  
 
print( "ANOVA results: F=", f_val, ", P =", p_val )
```

## 4wd and fwd

```python
f_val, p_val = stats.f_oneway(
    grouped_wheels.get_group('4wd')['price'], 
    grouped_wheels.get_group('fwd')['price']
)  
 
print( "ANOVA results: F=", f_val, ", P =", p_val )
```
